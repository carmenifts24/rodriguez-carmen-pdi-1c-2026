## 4. Curar el dataset

El modelo aprende a partir de los ejemplos que recibe. Si los ejemplos están sesgados, el modelo puede aprender atajos incorrectos.

Revisá estos criterios:

- **Balance de clases**: cada clase debería tener una cantidad similar de imágenes.
- **Variedad**: incluí distintos fondos, ángulos, tamaños e iluminaciones.
- **Ruido**: eliminá imágenes borrosas, duplicadas o irrelevantes.
- **Sesgo de fondo**: evitá que una clase aparezca siempre con el mismo fondo.

Ejemplo: si todas las fotos de gatos están en un sillón rojo y todas las de perros están en pasto verde, el modelo puede aprender a distinguir fondos en lugar de animales.

## 9. Cargar modelo y etiquetas

Esta celda espera que `keras_model.h5` y `labels.txt` estén en la misma carpeta del notebook.

````python
from pathlib import Path

import gradio as gr
import numpy as np
from PIL import Image, ImageOps
# Los modelos .h5 exportados por Teachable Machine usan el formato Keras 2.
# tf_keras conserva esa compatibilidad en el entorno TensorFlow actual.
from tf_keras.models import load_model

ruta_modelo = Path("keras_model.h5")
ruta_etiquetas = Path("labels.txt")

# Verificación temprana: fallar con un mensaje claro en lugar de un error
# críptico de HDF5 si los archivos no están en la carpeta del notebook.
if not ruta_modelo.exists():
    raise FileNotFoundError("No se encontró keras_model.h5. Revisá que el archivo esté en la carpeta del notebook.")

if not ruta_etiquetas.exists():
    raise FileNotFoundError("No se encontró labels.txt. Revisá que el archivo esté en la carpeta del notebook.")

# compile=False evita que Keras intente reconstruir el optimizador guardado
# (Adam con sus estadísticas de momento), que no es necesario para inferencia
# y puede causar errores de compatibilidad al cargar modelos de Teachable Machine.
modelo_teachable = load_model(ruta_modelo, compile=False)

with ruta_etiquetas.open("r", encoding="utf-8") as archivo_etiquetas:
    lineas_etiquetas = archivo_etiquetas.readlines()

nombres_clases = []

for linea in lineas_etiquetas:
    # labels.txt tiene el formato "0 nombre_clase\n"; split(" ", 1) divide
    # solo en el primer espacio y [-1] toma el nombre descartando el índice numérico.
    nombre_clase = linea.strip().split(" ", 1)[-1]
    nombres_clases.append(nombre_clase)

print("Clases cargadas:", nombres_clases)
````

## 10. Definir la función de predicción

Teachable Machine espera imágenes de 224 x 224 píxeles y valores normalizados al rango `[-1, 1]`. Por eso adaptamos cada imagen antes de enviarla al modelo.

````python
# Preprocesamiento e inferencia para modelos exportados desde Teachable Machine.
# preprocesar_imagen_para_teachable aplica el pipeline exacto que Teachable Machine
# usó durante el entrenamiento; predecir_imagen lo invoca y devuelve un dict
# {nombre_clase: probabilidad} compatible con la salida de Label de Gradio.

def preprocesar_imagen_para_teachable(imagen_numpy):
    imagen_pil = Image.fromarray(imagen_numpy.astype("uint8"), "RGB")

    tamano_esperado = (224, 224)
    # ImageOps.fit hace crop centrado + resize en un solo paso, preservando
    # la relación de aspecto sin deformar la imagen (a diferencia de resize directo).
    imagen_redimensionada = ImageOps.fit(
        imagen_pil,
        tamano_esperado,
        Image.Resampling.LANCZOS,  # filtro de alta calidad para reducción de escala
    )

    arreglo_imagen = np.asarray(imagen_redimensionada)
    arreglo_imagen = arreglo_imagen.astype(np.float32)
    # Normalización estándar de Teachable Machine: mapea [0, 255] → [-1, 1].
    arreglo_imagen = (arreglo_imagen / 127.5) - 1

    # El modelo espera un lote: shape (1, 224, 224, 3). expand_dims agrega el eje de batch.
    lote_imagenes = np.expand_dims(arreglo_imagen, axis=0)

    return lote_imagenes


def predecir_imagen(imagen_entrada):
    if imagen_entrada is None:
        return {"Sin imagen": 1.0}

    lote_imagenes = preprocesar_imagen_para_teachable(imagen_entrada)
    prediccion = modelo_teachable.predict(lote_imagenes, verbose=0)

    probabilidades = {}

    for indice, nombre_clase in enumerate(nombres_clases):
        # prediccion tiene shape (1, n_clases); [0] desempaqueta el único elemento del lote.
        puntaje = float(prediccion[0][indice])
        probabilidades[nombre_clase] = puntaje

    return probabilidades
````

**Consigna de lectura:** 

**identificá dónde se redimensiona la imagen y dónde se normaliza.**
- **Redimensión** → `ImageOps.fit(imagen_pil, (224, 224), ...)`: transforma la imagen a exactamente 224×224 píxeles recortando si es necesario, sin deformar las proporciones.
- **Normalización** → `(arreglo_imagen / 127.5) - 1`: convierte los píxeles del rango `[0, 255]` al rango `[-1, 1]`, que es el formato que espera MobileNet (la arquitectura base que usa Teachable Machine internamente).

**¿Por qué esos pasos deben coincidir con el entrenamiento original?**
Si durante el entrenamiento las imágenes se normalizaron al rango `[-1, 1]` y durante la predicción se normalizaran al rango `[0, 1]`, los valores de entrada serían completamente distintos a los que el modelo aprendió a interpretar. El resultado sería predicciones incorrectas aunque la imagen sea perfectamente nítida y relevante.

Este principio aplica a cualquier transformación: redimensionamiento, conversión de color, recorte, normalización. Todo lo que se hizo en el preprocesamiento de entrenamiento debe replicarse exactamente en la inferencia. En producción, este contrato entre entrenamiento y predicción es una de las fuentes más frecuentes de errores silenciosos.

## 11. Crear la interfaz con Gradio

La interfaz permite subir una imagen o usar la cámara, según el entorno donde ejecutes el notebook. El parámetro `share=True` genera un enlace temporal para probar el modelo desde otros dispositivos.

````python
demo = gr.Interface(
    fn=predecir_imagen,
    inputs=gr.Image(type="numpy"),
    outputs=gr.Label(num_top_classes=3),
    title="Clasificador de imágenes con Teachable Machine",
    description="Subí una imagen para probar el modelo entrenado en clase.",
)

# Se usa el servidor local para no depender del túnel público de Gradio.
demo.launch(share=False)
````
