### El Archivo app.py: Corazón del Space

El script principal debe denominarse estrictamente `app.py`. A continuación, analizamos una plantilla base didáctica que implementa un clasificador general de imágenes usando la arquitectura **Vision Transformer (ViT)**.

````python
# Ejemplo conceptual de app.py: muestra la estructura completa de una aplicación Gradio
# que expone un modelo de Deep Learning como servicio web interactivo.
# Sigue el patrón: carga del modelo → función de inferencia → interfaz → lanzamiento.

# Gradio: framework para construir UIs web interactivas con Python sin HTML/JS.

import gradio as gr
# transformers: librería de HuggingFace para cargar modelos de lenguaje y visión pre-entrenados.
from transformers import pipeline
# PIL (Pillow): librería estándar de manipulación de imágenes en Python.
from PIL import Image

print('✦ Inicializando la aplicación y cargando recursos de Deep Learning...')

# 1. CARGA DEL MODELO: Se ejecuta una única vez al levantar el contenedor para evitar sobrecostos en cada consulta.
#    pipeline() es una abstracción de alto nivel de HuggingFace: descarga automáticamente los pesos del modelo
#    y encapsula el preprocesamiento (resize, normalización) y la inferencia en una sola llamada.
#    'image-classification' indica la tarea; el pipeline elige internamente el head de clasificación correcto.
#    'google/vit-base-patch16-224': ViT (Vision Transformer) es una arquitectura que divide la imagen
#    en parches de 16x16 px y los procesa como tokens —igual que BERT con palabras—.
#    Fue entrenado con ImageNet-21k (21 millones de imágenes, 21 000 clases) y fine-tuned en ImageNet-1k.
#    'base' indica tamaño mediano; existe 'large' con más parámetros pero mayor costo computacional.
clasificador = pipeline(
    'image-classification',
    model='google/vit-base-patch16-224'
)
print('✓ Modelo Vision Transformer cargado y listo.')


# 2. FUNCIÓN DE NEGOCIO: Es el puente entre la UI y el modelo; Gradio la llama cada vez que el usuario envía una imagen.
def clasificar_imagen(imagen):
    """
    Recibe un objeto PIL.Image desde Gradio, lo pasa al modelo y devuelve
    un dict {clase: probabilidad} listo para renderizar en gr.Label.
    """
    # Gradio puede entregar None si el usuario pulsa Submit sin haber cargado imagen.
    if imagen is None:
        return {'Error': 'No se cargó ninguna imagen'}  # gr.Label espera un dict; retornar uno con clave 'Error' lo muestra en la UI.
        
    try:
        # El pipeline recibe el PIL.Image, lo redimensiona a 224x224 (tamaño esperado por ViT),
        # aplica normalización con la media/std de ImageNet, ejecuta la inferencia
        # y devuelve una lista de dicts: [{'label': 'cat', 'score': 0.92}, ...]
        # ordenada de mayor a menor probabilidad.
        resultados = clasificador(imagen)
        
        # gr.Label requiere un dict {etiqueta: score} con valores Python float nativos,
        # no numpy float32, por eso se aplica float() explícitamente.
        # [:5] limita a las 5 predicciones más probables (top-5), igual que la métrica estándar de ImageNet.
        predicciones_formateadas = {}
        for res in resultados[:5]:
            predicciones_formateadas[res['label']] = float(res['score'])
            
        return predicciones_formateadas
    except Exception as e:
        # Captura errores de inferencia (imagen corrupta, formato inesperado, etc.)
        # y los muestra en la UI en lugar de crashear la app.
        return {'Error en inferencia': str(e)}


# 3. INTERFAZ GRADIO: gr.Interface conecta automáticamente inputs → función → outputs
#    y genera el HTML, CSS y JS necesarios sin configuración adicional.
demo = gr.Interface(
    fn=clasificar_imagen,                                    # función a ejecutar al hacer Submit.
    inputs=gr.Image(type='pil', label='Suban una imagen'),  # type='pil' porque el pipeline de HuggingFace espera un objeto PIL, no un array numpy.
    outputs=gr.Label(num_top_classes=5, label='Predicciones de ImageNet'),  # gr.Label renderiza el dict como un gráfico de barras con las top-5 clases.
    title='Clasificador de Imágenes con ViT',
    description='Carguen una imagen para que el modelo estime su categoría conceptual de ImageNet.',
)

# 4. LANZAMIENTO DE LA APLICACIÓN
if __name__ == '__main__':  # Evita que el servidor se inicie si este archivo es importado como módulo en lugar de ejecutado directamente.
    print('✦ Iniciando el servidor local de Gradio...')
    # share=False (default): levanta el servidor solo en localhost.
    # En HuggingFace Spaces el servidor lo inicia el runtime del Space, no este bloque.
    demo.launch()
````

## 5. Arquitectura Didáctica de 3 Capas

Cuando diseñamos interfaces interactivas para modelos de Visión Digital, es fundamental estructurar el código siguiendo la arquitectura de tres capas para garantizar que sea modular y legible:

```
┌─────────────────────────────────────────────────────────────┐
│             PRESENTATION LAYER (Gradio UI)                 │ ← Interfaz gráfica de usuario
├─────────────────────────────────────────────────────────────┤
│             BUSINESS LOGIC (Funciones de negocio)          │ ← Validación, formateo y control
├─────────────────────────────────────────────────────────────┤
│             DATA LAYER (Modelos deep learning / APIs)       │ ← Carga de pesos e inferencia con PyTorch
└─────────────────────────────────────────────────────────────┘
```

A continuación, presentamos un ejemplo completo de código modular aplicando este principio estructural.

````python
import os
import gradio as gr
from transformers import pipeline
from PIL import Image

print('✦ Configurando las capas de la aplicación...')

# =====================================================================
# 1. DATA LAYER (Capa de Datos): Carga de Modelos e Inferencia
# =====================================================================
print('✦ [Data Layer] Cargando modelo preentrenado...')

# pipeline() descarga los pesos de HuggingFace Hub la primera vez y los cachea en disco.
# Se instancia aquí, fuera de cualquier función, para que la carga ocurra una sola vez
# al iniciar el servidor — no en cada llamada del usuario, lo que sería prohibitivo en latencia.
clasificador_modelo = pipeline(
    'image-classification',              # Tarea: clasifica una imagen en categorías de ImageNet.
    model='google/vit-base-patch16-224'  # ViT-Base: divide la imagen en parches 16×16 px y los
                                         # procesa como tokens —misma idea que BERT con palabras—.
                                         # Entrenado en ImageNet-21k, fine-tuned en ImageNet-1k.
)
print('✓ [Data Layer] Modelo cargado con éxito en memoria.')


# =====================================================================
# 2. BUSINESS LOGIC LAYER (Capa de Lógica de Negocio): Validaciones y Control
# =====================================================================
def procesar_y_clasificar(imagen):
    """
    Valida la entrada, llama al modelo y formatea la salida para gr.Label.
    Gradio invoca esta función cada vez que el usuario hace clic en el botón.
    """
    # Gradio entrega None cuando el usuario pulsa el botón sin haber cargado imagen.
    if imagen is None:
        print('✗ [Business Layer] Intento de clasificación sin datos de imagen.')
        return {'Error': 'Por favor, carguen una imagen válida en el visor.'}
        
    print('✦ [Business Layer] Muestra recibida. Ejecutando preprocesamiento e inferencia...')
    
    try:
        # El pipeline recibe PIL.Image, aplica internamente resize a 224×224
        # y normalización con media/std de ImageNet antes de la inferencia.
        # Devuelve lista ordenada: [{'label': 'tabby cat', 'score': 0.91}, ...]
        resultados = clasificador_modelo(imagen)
        
        # gr.Label espera {str: float}; los scores del pipeline son numpy.float32,
        # que Gradio no serializa correctamente — float() los convierte a Python nativo.
        # [:5] aplica el criterio top-5 estándar de ImageNet para reportar predicciones.
        salida_diccionario = {}
        for res in resultados[:5]:
            salida_diccionario[res['label']] = float(res['score'])
            
        print('✓ [Business Layer] Procesamiento y estimación completados.')
        return salida_diccionario
        
    except Exception as error:
        # Captura errores inesperados (imagen corrupta, OOM, etc.) y los expone
        # en la UI en lugar de crashear el servidor de Gradio.
        print(f'✗ [Business Layer] Falla en ejecución del pipeline: {error}')
        return {'Error de Inferencia': str(error)}


# =====================================================================
# 3. PRESENTATION LAYER (Capa de Presentación): Interfaz de Gradio (Blocks)
# =====================================================================
print('✦ [Presentation Layer] Construyendo interfaz web...')

# gr.Blocks es la API de bajo nivel de Gradio: permite layouts personalizados
# a diferencia de gr.Interface, que solo admite una columna de inputs y otra de outputs.
# El bloque `with` registra los componentes en el grafo interno de la app.
with gr.Blocks(theme=gr.themes.Soft()) as interfaz_app:
    gr.Markdown('# ✦ Clasificador de Imágenes de Elite (ViT)')
    gr.Markdown('Carguen una imagen en la columna izquierda y hagan clic en **Clasificar muestra**.')
    
    # gr.Row() y gr.Column() son contenedores de layout CSS flexbox;
    # no gestionan recursos — solo organizan los componentes en filas y columnas.
    with gr.Row():
        with gr.Column():
            # type='pil' hace que Gradio entregue un PIL.Image a la función,
            # en lugar de un numpy array — necesario porque el pipeline de HuggingFace
            # aplica su propio preprocesamiento a objetos PIL.
            imagen_input = gr.Image(type='pil', label='Muestra Visual (Entrada)')
            # variant='primary' aplica el color de acento del tema al botón.
            boton_ejecutar = gr.Button('Clasificar muestra', variant='primary')
            
        with gr.Column():
            # gr.Label renderiza el dict {clase: score} como gráfico de barras horizontal;
            # num_top_classes=5 limita cuántas barras se muestran aunque el dict tenga más entradas.
            etiqueta_output = gr.Label(num_top_classes=5, label='Estimaciones de Confianza (Salida)')
            
    # .click() registra el evento: cuando el usuario presiona el botón,
    # Gradio toma el valor actual de imagen_input, lo pasa a procesar_y_clasificar
    # y escribe el resultado en etiqueta_output — todo vía WebSocket sin recargar la página.
    boton_ejecutar.click(
        fn=procesar_y_clasificar,
        inputs=imagen_input,
        outputs=etiqueta_output
    )
    
print('✓ [Presentation Layer] UI inicializada correctamente.')
````

## 1. ¿Por qué separar la carga del modelo de la función de procesamiento?

**Razón computacional:**
`pipeline()` descarga los pesos del modelo desde HuggingFace Hub (~350 MB para `vit-base-patch16-224`) y los carga en RAM/VRAM. Si esta operación estuviera dentro de `procesar_y_clasificar()`, se ejecutaría **en cada llamada del usuario**, multiplicando la latencia por el número de consultas. Al instanciarlo una sola vez al arrancar el servidor, el costo se paga una sola vez.

**Razón didáctica:**
Refleja el patrón de arquitectura en capas: la **Data Layer** es responsable de los recursos persistentes (el modelo en memoria), mientras que la **Business Logic Layer** solo orquesta el flujo de datos. Un estudiante que lee el código puede identificar inmediatamente qué es infraestructura y qué es lógica de negocio.

## 2. `ModuleNotFoundError` en Hugging Face Spaces

El problema reside en que el paquete que falta **no está declarado en `requirements.txt`**. HuggingFace Spaces construye el entorno instalando exclusivamente las dependencias listadas en ese archivo; si una biblioteca se usa en el código pero no aparece ahí, el contenedor no la instala y Python lanza `ModuleNotFoundError` al intentar importarla.

**Solución:** agregar la dependencia faltante a `requirements.txt`, por ejemplo:

transformers>=4.35.0
torch>=2.0.0
pillow>=9.0.0
gradio>=4.0.0
timm>=0.9.0        # ← agregar si el error es sobre 'timm'

El Space re-ejecuta `pip install -r requirements.txt` en el siguiente deploy.
