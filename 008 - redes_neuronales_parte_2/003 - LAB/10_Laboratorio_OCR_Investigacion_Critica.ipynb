## Terminología clave (Microglosario)

* **OCR (Reconocimiento Óptico de Caracteres):** Técnica que permite extraer texto a partir de imágenes, escaneos o fotografías de documentos.
* **Fuente primaria:** Recurso técnico original donde una afirmación puede verificarse directamente, por ejemplo la documentación oficial, el repositorio de un modelo o un paper.
* **Handle de modelo:** Identificador exacto con el que un modelo se publica en plataformas como ✦ 🤗 Hugging Face.
* **Pipeline:** Abstracción de `transformers` que simplifica la carga del modelo, el preprocesamiento y la inferencia.
* **CER / WER:** Métricas de error para comparar el texto reconocido con un texto de referencia. CER mide error por carácter y WER por palabra.
* **Trazabilidad:** Registro explícito de qué consultaron, qué fuente usaron, qué verificaron y qué decisión tomaron.

## Consigna de Lectura e Interpretación

Mientras verifican, deténganse en estas preguntas:

1. **¿Todas las afirmaciones equivocadas son igual de graves para un desarrollo real?**

No. La gravedad depende de qué tan rápido falla y qué tan difícil es detectar el error:

Error en la sintaxis del pipeline (result['generated_text'] sin índice): falla inmediatamente con TypeError en la primera ejecución. Es grave pero fácil de detectar y corregir.
Handle inexistente (baidu-nlp/clova-ocr-vietnamese): también falla rápido con un 404 o RepositoryNotFoundError. Grave pero visible.
Error de arquitectura (ViT + GPT-2 en vez de BEiT + RoBERTa): no rompe nada en ejecución, pero contamina la documentación técnica, los argumentos de selección de modelo y cualquier análisis de limitaciones. Es el más peligroso a largo plazo porque puede pasar desapercibido indefinidamente.
La jerarquía real es: los errores que rompen el código se detectan solos; los errores conceptuales que no rompen nada se propagan.

2. **¿Qué diferencia entre incorrecto, impreciso y no verificable?**

Incorrecto: la afirmación contradice la evidencia directa de una fuente primaria. Ejemplo: decir que el handle baidu-nlp/clova-ocr-vietnamese existe cuando devuelve 404. Hay una respuesta definitiva: es falso.

Impreciso: la afirmación es verdadera en parte pero omite condiciones que cambian el comportamiento real. Ejemplo: decir que TrOCR funciona para OCR en español sin aclarar que fue entrenado en inglés. No es mentira, pero lleva a expectativas incorrectas sobre el rendimiento real.

No verificable: no existe una fuente primaria accesible que permita confirmarla ni refutarla. Ejemplo: afirmar que un modelo "es el más usado en producción" sin citar métricas ni encuestas públicas. No se puede auditar con evidencia directa.

La diferencia práctica: lo incorrecto se puede corregir con certeza; lo impreciso requiere pruebas propias para entender el alcance real; lo no verificable obliga a asumir un riesgo desconocido.

3. **Si un handle está mal escrito pero la idea general parece razonable, ¿alcanza para implementarlo?**

No. El handle es exactamente la pieza que conecta el código con el modelo real. Un handle incorrecto produce uno de estos tres resultados: error 404 (modelo no encontrado), carga un modelo distinto al esperado sin aviso, o carga una versión desactualizada o de otro autor con diferente comportamiento.

El caso más peligroso es el segundo: si existe otro repositorio con un nombre similar, el pipeline lo carga silenciosamente y el código corre sin errores, pero las predicciones corresponden a un modelo que no elegiste, con licencia que no revisaste y métricas que no conocés. La "idea general razonable" no reemplaza la verificación del identificador exacto porque en sistemas de software el handle no es una descripción aproximada — es una dirección única. Estar cerca no es suficiente.
