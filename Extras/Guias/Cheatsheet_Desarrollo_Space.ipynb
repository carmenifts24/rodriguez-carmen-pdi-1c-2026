## ✦ Componentes Esenciales de Gradio

La API de Gradio cuenta con componentes prediseñados y configurados para interactuar de forma intuitiva con imágenes.

````python
# Comentario general: este bloque importa las bibliotecas que se usaran mas adelante; centralizar las importaciones deja claro que herramientas participan: manejo de datos, visualizacion, redes neuronales, imagenes o despliegue.
import gradio as gr  # Importa gradio para usar sus funciones en el notebook.

# 1. COMPONENTES DE ENTRADA (Inputs)
gr.Image(type='pil')           # Entrada visual que pasa una imagen PIL al backend
gr.Textbox()                   # Campo de entrada de texto alfanumérico
gr.Number()                    # Caja de entrada numérica
gr.Slider(minimum=0, maximum=1) # Deslizador lineal analítico

# 2. COMPONENTES DE SALIDA (Outputs)
gr.Label()                     # Caja de clasificación con barras de confianza
gr.Image()                     # Renderizador visual para imágenes procesadas o recortadas
gr.Textbox()                   # Caja de salida textual para metadatos o coordenadas

# 3. ESTRUCTURAS DE DISEÑO (Layout)
gr.Interface()                 # Abstracción simplificada y rápida de UI
gr.Blocks()                    # Contenedor avanzado para flujos y diseños a medida
gr.Row()                       # Organización horizontal (fila)
gr.Column()                    # Organización vertical (columna)
````

## ✦ Patrón de 3 Capas Modular

Este es el patrón arquitectónico estructurado que aplicamos para garantizar la separación de responsabilidades y la legibilidad de nuestro software.

````python
# Comentario general: este bloque importa las bibliotecas que se usaran mas adelante; centralizar las importaciones deja claro que herramientas participan: manejo de datos, visualizacion, redes neuronales, imagenes o despliegue.
import gradio as gr  # Importa gradio para usar sus funciones en el notebook.
from transformers import pipeline  # Importa componentes especificos desde transformers.

# =====================================================================
# 1. DATA LAYER: Carga única del modelo en memoria
# =====================================================================
modelo_vit = pipeline('image-classification', model='google/vit-base-patch16-224')  # Calcula y guarda el valor en la variable modelo_vit.

# =====================================================================
# 2. BUSINESS LOGIC: Validación y lógica de control
# =====================================================================
def procesar_muestra(imagen):  # Define la funcion procesar_muestra para reutilizar esta logica mas adelante.
    if imagen is None:  # Evalua una condicion para decidir que rama del codigo ejecutar.
        return {'Error': 'Por favor, carguen una muestra.'}  # Devuelve el resultado calculado por la funcion.
        
    try:  # Inicia un bloque protegido para manejar posibles errores.
        resultados = modelo_vit(imagen)  # Calcula y guarda el valor en la variable resultados.
        salida_limpia = {}  # Calcula y guarda el valor en la variable salida_limpia.
        for res in resultados[:5]:  # Inicia un bucle para recorrer elementos de una coleccion.
            salida_limpia[res['label']] = float(res['score'])  # Completa una llamada o estructura iniciada en lineas anteriores.
        return salida_limpia  # Devuelve el resultado calculado por la funcion.
    except Exception as e:  # Captura un error posible y define como responder ante el.
        return {'Error de Inferencia': str(e)}  # Devuelve el resultado calculado por la funcion.

# =====================================================================
# 3. PRESENTATION LAYER: Interfaz gráfica declarada con Blocks
# =====================================================================
with gr.Blocks(theme=gr.themes.Soft()) as demo_app:  # Abre un contexto controlado para manejar recursos de forma segura.
    gr.Markdown('# Clasificador Didáctico ViT')  # Completa una llamada o estructura iniciada en lineas anteriores.
    
    with gr.Row():  # Abre un contexto controlado para manejar recursos de forma segura.
        img_input = gr.Image(type='pil', label='Imagen de Entrada')  # Calcula y guarda el valor en la variable img_input.
        lbl_output = gr.Label(label='Predicciones de Salida')  # Calcula y guarda el valor en la variable lbl_output.
        
    btn_ejecutar = gr.Button('Procesar', variant='primary')  # Calcula y guarda el valor en la variable btn_ejecutar.
    btn_ejecutar.click(fn=procesar_muestra, inputs=img_input, outputs=lbl_output)  # Completa una llamada o estructura iniciada en lineas anteriores.

if __name__ == '__main__':  # Evalua una condicion para decidir que rama del codigo ejecutar.
    demo_app.launch()  # Completa una llamada o estructura iniciada en lineas anteriores.
````
