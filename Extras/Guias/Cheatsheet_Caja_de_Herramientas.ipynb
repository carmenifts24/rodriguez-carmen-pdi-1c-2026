## 1. Recetario de Capas de Keras y Arquitecturas

A continuación, presentamos la sintaxis de las capas fundamentales para armar modelos de Perceptrón Multicapa (MLP) y Redes Neuronales Convolucionales (CNNs).

```python
# Comentario general: este bloque importa las bibliotecas que se usaran mas adelante; centralizar las importaciones deja claro que herramientas participan: manejo de datos, visualizacion, redes neuronales, imagenes o despliegue.
import tensorflow as tf  # Importa tensorflow para usar sus funciones en el notebook.

# =========================================================================
# RECETA 1: ARQUITECTURA DE PERCEPTRÓN MULTICAPA (MLP)
# =========================================================================

modelo_mlp = tf.keras.Sequential([  # Construye un modelo secuencial capa por capa.
    # Capa de Entrada: Especificamos la forma de la imagen (ej: 28x28 y 1 canal)
    tf.keras.layers.Input(shape=(28, 28, 1)),  # Ejecuta esta instruccion como parte del flujo principal del notebook.
    
    # Capa de Aplanado: Redimensiona de 2D (28x28) a vector plano (784)
    tf.keras.layers.Flatten(),  # Aplana las caracteristicas para conectarlas con capas densas.
    
    # Capas ocultas totalmente conectadas (Dense) con activación ReLU
    tf.keras.layers.Dense(units=64, activation='relu'),  # Agrega una capa densa para combinar caracteristicas y producir una salida.
    tf.keras.layers.Dense(units=32, activation='relu'),  # Agrega una capa densa para combinar caracteristicas y producir una salida.
    
    # Capa de salida: una neurona por clase y activación Softmax para probabilidades
    tf.keras.layers.Dense(units=10, activation='softmax')  # Agrega una capa densa para combinar caracteristicas y producir una salida.
])  # Completa una llamada o estructura iniciada en lineas anteriores.

# =========================================================================
# RECETA 2: ARQUITECTURA DE RED NEURONAL CONVOLUCIONAL (CNN)
# =========================================================================

modelo_cnn = tf.keras.Sequential([  # Construye un modelo secuencial capa por capa.
    # Capa de Entrada: Imagen 2D sin aplanar
    tf.keras.layers.Input(shape=(28, 28, 1)),  # Ejecuta esta instruccion como parte del flujo principal del notebook.
    
    # Primera capa convolucional: 16 filtros de 3x3 y activación ReLU
    tf.keras.layers.Conv2D(filters=16, kernel_size=(3, 3), activation='relu'),  # Agrega una capa convolucional para extraer patrones espaciales de la imagen.
    # Primera capa de Pooling: reduce resolución a la mitad espacial
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),  # Reduce la resolucion del mapa de caracteristicas conservando respuestas fuertes.
    
    # Segunda capa convolucional: 32 filtros de 3x3 y activación ReLU
    tf.keras.layers.Conv2D(filters=32, kernel_size=(3, 3), activation='relu'),  # Agrega una capa convolucional para extraer patrones espaciales de la imagen.
    # Segunda capa de Pooling: reduce resolución espacial a la mitad
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),  # Reduce la resolucion del mapa de caracteristicas conservando respuestas fuertes.
    
    # Aplanamos los mapas de características jerárquicos reducidos
    tf.keras.layers.Flatten(),  # Aplana las caracteristicas para conectarlas con capas densas.
    
    # Capa Dense intermedia para razonamiento abstracto
    tf.keras.layers.Dense(units=64, activation='relu'),  # Agrega una capa densa para combinar caracteristicas y producir una salida.
    
    # Capa de salida: probabilidades para las 10 categorías
    tf.keras.layers.Dense(units=10, activation='softmax')  # Agrega una capa densa para combinar caracteristicas y producir una salida.
])  # Completa una llamada o estructura iniciada en lineas anteriores.

print("✓ Plantillas de arquitecturas MLP y CNN instanciadas correctamente.")  # Muestra informacion en consola para verificar el avance o los resultados.
```

## 2. Compilación y Configuración del Aprendizaje

La compilación define de qué manera va a aprender la red durante su optimización numérica.

```python
# Comentario general: este bloque configura el entrenamiento del modelo; el optimizador indica como ajustar los pesos y la funcion de perdida mide que tan lejos estan las predicciones de las respuestas esperadas.
# =========================================================================
# CONFIGURACIÓN DE COMPILACIÓN ESTÁNDAR
# =========================================================================

modelo_cnn.compile(  # Configura el optimizador, la perdida y las metricas del modelo.
    # Optimizador Adam: ajusta dinámicamente la tasa de aprendizaje
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),  # Calcula y guarda el valor en la variable optimizer.
    
    # Loss para problemas multiclase donde las etiquetas son números enteros (0, 1, 2...)
    loss='sparse_categorical_crossentropy',  # Calcula y guarda el valor en la variable loss.
    
    # Métrica para evaluar visualmente el rendimiento
    metrics=['accuracy']  # Referencia la precision como metrica de desempeno del modelo.
)  # Completa una llamada o estructura iniciada en lineas anteriores.

print("✓ Configuración de aprendizaje y optimización compilada exitosamente.")  # Muestra informacion en consola para verificar el avance o los resultados.
```

## 3. Receta para Gráficas de Curvas de Aprendizaje

Una función útil para dibujar rápidamente las curvas de pérdida y exactitud a partir del historial retornado por `model.fit()`.

```python
# Comentario general: este bloque importa las bibliotecas que se usaran mas adelante; centralizar las importaciones deja claro que herramientas participan: manejo de datos, visualizacion, redes neuronales, imagenes o despliegue.
import matplotlib.pyplot as plt  # Importa matplotlib para usar sus funciones en el notebook.

def graficar_curvas_aprendizaje(historial_entrenamiento):  # Define la funcion graficar_curvas_aprendizaje para reutilizar esta logica mas adelante.
    """
    Dibuja de forma elegante y automatizada las métricas de precisión y pérdida.
    """
    historial_dict = historial_entrenamiento.history  # Calcula y guarda el valor en la variable historial_dict.
    epocas = range(1, len(historial_dict['accuracy']) + 1)  # Referencia la precision como metrica de desempeno del modelo.
    
    plt.figure(figsize=(12, 5))  # Crea una figura para graficar resultados o imagenes.
    
    # Dibujamos la Precisión
    plt.subplot(1, 2, 1)  # Selecciona la posicion del grafico dentro de una grilla.
    plt.plot(epocas, historial_dict['accuracy'], 'bo-', label='Precisión')  # Configura o dibuja una visualizacion para interpretar los datos.
    if 'val_accuracy' in historial_dict:  # Evalua una condicion para decidir que rama del codigo ejecutar.
        plt.plot(epocas, historial_dict['val_accuracy'], 'ro-', label='Precisión Val')  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.title('Precisión durante el Aprendizaje', fontsize=12, fontweight="bold")  # Asigna un titulo descriptivo al grafico.
    plt.xlabel('Épocas')  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.ylabel('Precisión (Accuracy)')  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.legend()  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.grid(True, linestyle="--", alpha=0.5)  # Configura o dibuja una visualizacion para interpretar los datos.
    
    # Dibujamos la Pérdida
    plt.subplot(1, 2, 2)  # Selecciona la posicion del grafico dentro de una grilla.
    plt.plot(epocas, historial_dict['loss'], 'gs-', label='Pérdida')  # Configura o dibuja una visualizacion para interpretar los datos.
    if 'val_loss' in historial_dict:  # Evalua una condicion para decidir que rama del codigo ejecutar.
        plt.plot(epocas, historial_dict['val_loss'], 'rs-', label='Pérdida Val')  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.title('Evolución de la Pérdida', fontsize=12, fontweight="bold")  # Asigna un titulo descriptivo al grafico.
    plt.xlabel('Épocas')  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.ylabel('Pérdida (Loss)')  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.legend()  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.grid(True, linestyle="--", alpha=0.5)  # Configura o dibuja una visualizacion para interpretar los datos.
    
    plt.tight_layout()  # Configura o dibuja una visualizacion para interpretar los datos.
    plt.show()  # Renderiza y muestra la visualizacion completa.

print("✓ Función 'graficar_curvas_aprendizaje' compilada y disponible.")  # Muestra informacion en consola para verificar el avance o los resultados.
```
